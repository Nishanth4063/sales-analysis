```python
import pandas as pd
df = pd.read_csv("sales_data.csv")
print(df.shape, df.isna().sum().sum(), df.duplicated().sum())
print(df["Sales_Amount"].sum())
print(df.groupby("Region")["Sales_Amount"].sum().round(2))
```

```python
print(df.groupby("Product_Category")["Sales_Amount"].sum().round(2).sort_values(ascending=False))
print(df.groupby("Sales_Rep")["Sales_Amount"].sum().round(2).sort_values(ascending=False))
print(df.groupby("Sales_Channel")["Sales_Amount"].sum().round(2))
```

```python
import matplotlib.pyplot as plt
df["Sale_Date"] = pd.to_datetime(df["Sale_Date"])
m = df[df["Sale_Date"].dt.year == 2023].groupby(df["Sale_Date"].dt.month)["Sales_Amount"].sum()
m.plot(kind="line", marker="o", title="Monthly Sales (2023)")
plt.show()
```
